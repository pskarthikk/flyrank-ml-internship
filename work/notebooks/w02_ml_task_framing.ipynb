## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
ml_task_type = "Scoring / Ranking"

print("ML Task Type:", ml_task_type)
```

The target is whether a content page should be prioritized for refresh.

Since there is no direct "needs refresh" label, a proxy target can be created using historical performance trends. Pages showing consistent declines in important performance metrics can be considered higher priority for refresh.

The model supports editorial decision making rather than making automatic publishing decisions.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Shape:", df.shape)

df.head()
```

A suitable success metric is **Precision@K**.

Editors usually have limited time, so the model should rank the most valuable refresh opportunities at the top. Precision@K measures how many of the highest-ranked recommendations are actually useful, making it appropriate for prioritization.

```python
success_metric = "Precision@K"

print("Success Metric:", success_metric)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule cannot easily capture the combined effects of many different features that influence content performance.

Machine learning can learn complex patterns from historical data and combine multiple signals when estimating refresh priority. This allows editors to focus on the pages that are most likely to benefit from updates instead of relying on simple thresholds.
