```python
from pathlib import Path
import sys

repo = Path("/content/flyrank-ml-internship")

# Make the repository scripts importable.
sys.path.insert(0, str(repo / "scripts"))

from ml_utils import (
    MODEL_NUMERIC_FEATURES,
    MODEL_CATEGORICAL_FEATURES,
)

import numpy as np
import pandas as pd

RAW_PATH = repo / "data/raw/content_refresh_anonymized.csv"
FEATURE_PATH = repo / "data/processed/refresh_feature_vector.csv"

# Load raw data
df = pd.read_csv(RAW_PATH)

# Create the target.
# This is the outcome we want to predict, not a model feature.
df["is_declining_label"] = (
    df["trend_direction"] == "down"
).astype(int)

# Create the log features required by ml_utils.py.
for source, target in [
    ("impressions_90d", "log_impressions_90d"),
    ("clicks_90d", "log_clicks_90d"),
    ("sessions_90d", "log_sessions_90d"),
    ("ai_sessions_90d", "log_ai_sessions_90d"),
]:
    values = pd.to_numeric(df[source], errors="coerce").fillna(0)
    df[target] = np.log1p(values)

# Verify all official model features exist.
required_features = (
    MODEL_NUMERIC_FEATURES
    + MODEL_CATEGORICAL_FEATURES
)

missing_features = [
    col for col in required_features
    if col not in df.columns
]

print("Rows:", len(df))
print("Target positive rows:", int(df["is_declining_label"].sum()))
print("Target positive rate:", round(df["is_declining_label"].mean(), 4))
print("Missing official model features:", missing_features)

assert not missing_features

# Keep the fields needed for modeling and grouped validation.
keep_columns = [
    "content_id",
    "client_id",
    "is_declining_label",
] + required_features

feature_df = df[keep_columns].copy()

# Save the prepared feature vector.
FEATURE_PATH.parent.mkdir(parents=True, exist_ok=True)
feature_df.to_csv(FEATURE_PATH, index=False)

print("\nFeature vector written to:")
print(FEATURE_PATH)

print("\nFeature vector shape:", feature_df.shape)
print("Numeric features:", len(MODEL_NUMERIC_FEATURES))
print("Categorical features:", len(MODEL_CATEGORICAL_FEATURES))
```

```python
import numpy as np
import pandas as pd

# Load the prepared feature vector.
feature_df = pd.read_csv(
    "/content/flyrank-ml-internship/data/processed/refresh_feature_vector.csv"
)

RANDOM_STATE = 42

# Keep client_id for grouping, but never use it as a model feature.
client_series = feature_df["client_id"].fillna("unknown").astype(str)

unique_clients = client_series.drop_duplicates().to_numpy()

rng = np.random.default_rng(RANDOM_STATE)
shuffled_clients = rng.permutation(unique_clients)

test_client_count = max(
    1,
    int(round(len(shuffled_clients) * 0.20))
)

test_clients = set(shuffled_clients[:test_client_count])

test_mask = client_series.isin(test_clients)

train_indices = np.flatnonzero(~test_mask.to_numpy())
test_indices = np.flatnonzero(test_mask.to_numpy())

y = feature_df["is_declining_label"].astype(int)

print("Total clients:", len(unique_clients))
print("Held-out clients:", len(test_clients))
print("Training clients:", len(unique_clients) - len(test_clients))

print("\nTrain rows:", len(train_indices))
print("Test rows:", len(test_indices))

print("\nTrain declining rate:", round(y.iloc[train_indices].mean(), 4))
print("Test declining rate:", round(y.iloc[test_indices].mean(), 4))

print("\nTrain classes:", sorted(y.iloc[train_indices].unique().tolist()))
print("Test classes:", sorted(y.iloc[test_indices].unique().tolist()))

print("\nHeld-out client IDs:")
print(sorted(test_clients))

assert len(train_indices) > 0
assert len(test_indices) > 0
assert y.iloc[train_indices].nunique() == 2
assert y.iloc[test_indices].nunique() == 2

print("\nClient-holdout validation: PASS")
```

## 2. Split design

*Grouped by client? Time-aware? Say why this split is honest for your question.*

### Split design

I use a **client-holdout split** so that pages from the same client do not appear in both training and test data. This tests whether the model can generalize to clients it did not see during training.

The split uses a fixed random seed of 42 and holds out 20% of the 32 clients. This produces 27,675 training rows and 2,325 test rows. Both subsets contain both target classes.

The Week-4 baseline and all learned models will be evaluated on these same 2,325 held-out rows.

```python
# Section 2 — Split design

print("Validation design: client holdout")
print("Random seed: 42")
print("Total clients:", len(unique_clients))
print("Training clients:", len(unique_clients) - len(test_clients))
print("Held-out clients:", len(test_clients))
print("Training rows:", len(train_indices))
print("Test rows:", len(test_indices))
print("Training declining rate:", round(y.iloc[train_indices].mean(), 4))
print("Test declining rate:", round(y.iloc[test_indices].mean(), 4))

assert len(train_indices) == 27675
assert len(test_indices) == 2325
assert y.iloc[train_indices].nunique() == 2
assert y.iloc[test_indices].nunique() == 2

print("\nSplit validation: PASS")
print("Client IDs are retained only for grouping and are not model features.")
```
