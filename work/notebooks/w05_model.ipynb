```python
import subprocess
from pathlib import Path

repo_url = "https://github.com/pskarthikk/flyrank-ml-internship.git"
repo_path = Path("/content/flyrank-ml-internship")

if repo_path.exists():
    print("Repository already exists:", repo_path)
else:
    result = subprocess.run(
        ["git", "clone", repo_url, str(repo_path)],
        capture_output=True,
        text=True
    )

    print(result.stdout)
    if result.stderr:
        print(result.stderr)

print("\nRepo exists:", repo_path.exists())
```
