## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
from IPython.display import Markdown, display

display(Markdown("""
### Research Question

**Research Question:** Which content pages should be prioritized for review so that editors can focus on the pages with the highest potential for improvement?

**Decision:** The project helps determine which pages should be reviewed first instead of reviewing every page manually.

**Who acts on the output?** Content editors, SEO specialists, or website managers.

**Action:** Review the highest-ranked pages, update or improve their content, and monitor their performance over time.

**Cost of a wrong recommendation:** If a low-priority page is ranked too high, valuable review time is wasted. If an important page is missed, it may continue to lose visibility, traffic, or engagement before anyone notices.

**Why machine learning?** Simple rules cannot capture all the complex relationships between search performance, content quality, engagement, and freshness. Machine learning can combine many signals to produce a better ranking that supports human decision-making.
"""))
```
